```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("dataset/car data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDescriptive Statistics:")
display(df.describe())
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns.tolist())

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
df.columns = df.columns.str.strip()

print("Cleaned Column Names:")
print(df.columns.tolist())
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Numerical Features:")
print(numerical_columns)

display(df[numerical_columns].describe())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Selling_Price"], kde=True)

plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Frequency")

plt.show()
```

```python
df.columns = df.columns.str.strip()

print("Column names:")
for col in df.columns:
    print(repr(col))
```

```python
data = df.copy()

# Remove Car_Name if it exists
if "Car_Name" in data.columns:
    data = data.drop("Car_Name", axis=1)

# Automatically find categorical columns
categorical_columns = data.select_dtypes(include="object").columns.tolist()

print("Categorical columns:")
print(categorical_columns)

# Convert categorical columns into numerical columns
data = pd.get_dummies(
    data,
    columns=categorical_columns,
    drop_first=True
)

print("\nPreprocessing completed!")
print("New dataset shape:", data.shape)

display(data.head())
```

```python
print("Processed columns:")
print(data.columns.tolist())

print("\nData types:")
print(data.dtypes)
```

```python
X = data.drop("Selling_Price", axis=1)
y = data["Selling_Price"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42
    )
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name} trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

results = []

for name, model in models.items():
    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R² Score": r2
    })

results_df = pd.DataFrame(results)

display(results_df)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="R² Score"
)

plt.title("Model Performance Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("R² Score")
plt.ylim(0, 1)
plt.xticks(rotation=15)

plt.show()
```

```python
best_model_name = results_df.loc[
    results_df["R² Score"].idxmax(),
    "Model"
]

best_model = models[best_model_name]

best_r2 = results_df.loc[
    results_df["R² Score"].idxmax(),
    "R² Score"
]

print("Best Model:", best_model_name)
print(f"Best R² Score: {best_r2:.4f}")
```

```python
best_predictions = best_model.predict(X_test)

plt.figure(figsize=(8, 6))

plt.scatter(y_test, best_predictions)

plt.xlabel("Actual Selling Price")
plt.ylabel("Predicted Selling Price")
plt.title(f"Actual vs Predicted Prices - {best_model_name}")

plt.show()
```

```python
# Select the first car from the test dataset
sample_car = X_test.iloc[[0]]

# Predict its selling price
sample_prediction = best_model.predict(sample_car)

print("Best Model:", best_model_name)
print(f"Predicted Selling Price: {sample_prediction[0]:.2f}")
print(f"Actual Selling Price: {y_test.iloc[0]:.2f}")
```

```python
comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": best_predictions
})

comparison["Difference"] = (
    comparison["Actual Price"] - comparison["Predicted Price"]
)

display(comparison.head(10))
```
