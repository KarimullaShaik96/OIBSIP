```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("dataset/Unemployment in India.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Dataset Shape:", df.shape)
print(df.columns.tolist())
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
# Remove rows containing missing values
df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nNew dataset shape:", df.shape)
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Convert Date column
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

print("Date conversion successful!")
print(df["Date"].head())
```

```python
df.info()
```

```python
region_avg = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False)

print("Average Unemployment Rate by Region:")
print(region_avg)
```

```python
plt.figure(figsize=(12, 6))

region_avg.plot(kind="bar")

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Region")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### Observation

The bar chart shows the average unemployment rate across different regions.
Regions with higher average unemployment rates experienced greater levels of
unemployment during the period covered by the dataset. The differences between
regions indicate that unemployment was not evenly distributed across India.

```python
print(df.columns.tolist())
```

```python
# Calculate monthly average unemployment rate
monthly_trend = (
    df.groupby(df["Date"].dt.to_period("M"))["Estimated Unemployment Rate (%)"]
    .mean()
)

# Convert period to string for plotting
monthly_trend.index = monthly_trend.index.astype(str)

print("Monthly Average Unemployment Rate:")
print(monthly_trend)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_trend.index,
    monthly_trend.values,
    marker="o"
)

plt.title("Monthly Unemployment Rate Trend")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

## Monthly Unemployment Trend

The line chart shows how the average unemployment rate changed over time.
The trend helps identify periods when unemployment increased or decreased.
Changes during 2020 are particularly important because they coincide with the
COVID-19 pandemic period.

```python
# Select three regions for comparison
selected_regions = df["Region"].unique()[:3]

print("Selected Regions:")
print(selected_regions)
```

```python
plt.figure(figsize=(14, 6))

for region in selected_regions:
    region_data = df[df["Region"] == region].sort_values("Date")
    
    plt.plot(
        region_data["Date"],
        region_data["Estimated Unemployment Rate (%)"],
        marker="o",
        label=region
    )

plt.title("Unemployment Rate Trend for Three Regions")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

## Regional Unemployment Comparison

The line chart compares unemployment-rate trends for three selected regions.
The unemployment rate varies over time, and the comparison helps identify
similarities and differences in regional unemployment patterns.

The sharp changes observed around 2020 are particularly important when
examining the effect of the COVID-19 pandemic on employment conditions.

```python
# Select numerical columns for correlation analysis
corr_cols = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]

correlation = df[corr_cols].corr()

print("Correlation Matrix:")
display(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Employment Indicators")
plt.tight_layout()
plt.show()
```

## Correlation Analysis

The correlation heatmap shows the relationships between unemployment rate,
estimated employment, and labour participation rate.

A positive correlation indicates that two variables tend to increase together,
while a negative correlation indicates that one variable tends to decrease when
the other increases.

The correlation values help identify relationships among the major employment
indicators in the dataset.

```python
import numpy as np

# Create COVID period classification
df["Period"] = np.where(
    df["Date"] < pd.Timestamp("2020-03-01"),
    "Pre-COVID",
    "COVID Period"
)

print(df[["Date", "Period"]].head())
```

```python
period_avg = (
    df.groupby("Period")["Estimated Unemployment Rate (%)"]
    .mean()
)

print("Average Unemployment Rate:")
print(period_avg)
```

```python
plt.figure(figsize=(8, 5))

period_avg.plot(kind="bar")

plt.title("Pre-COVID vs COVID-Period Unemployment Rate")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
pre_covid = period_avg["Pre-COVID"]
covid_period = period_avg["COVID Period"]

change = covid_period - pre_covid
percentage_change = (change / pre_covid) * 100

print(f"Pre-COVID Average: {pre_covid:.2f}%")
print(f"COVID-Period Average: {covid_period:.2f}%")
print(f"Change: {change:.2f} percentage points")
print(f"Percentage Change: {percentage_change:.2f}%")
```

## Pre-COVID vs COVID-Period Analysis

The unemployment rate was compared between the pre-COVID period and the
COVID-19 period. The comparison helps identify the impact of the pandemic on
unemployment levels.

The average unemployment rate increased/decreased during the COVID period
compared with the pre-COVID period, indicating the change in employment
conditions during the pandemic.
